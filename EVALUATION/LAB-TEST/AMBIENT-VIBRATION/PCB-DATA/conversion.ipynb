```python
import scipy.io
import numpy as np
import pandas as pd

# Step 1: Load .mat file
mat_data = scipy.io.loadmat('amb2.mat')
```

```python
# Step 2: Extract 'tdata' matrix
ambient_vib_pcb = mat_data['tdata']  # shape: (614400, 12)
print(f"'ambient_vib_pcb' shape: {ambient_vib_pcb.shape}")

```

```python
# Step 3: Define channel labels (optional but helpful)
sensor_labels = ['S1', 'S2', 'S3', 'S4']
axis_labels = ['X', 'Y', 'Z']
column_names = [f"{sensor}_{axis}" for sensor in sensor_labels for axis in axis_labels]

```

```python
# Step 4: Convert to DataFrame and add column names
df = pd.DataFrame(ambient_vib_pcb, columns=column_names)

```

```python
# Step 5: Save to .npy
np.save('ambient_vib_pcb.npy', ambient_vib_pcb)
print("[Saved] 'ambient_vib_pcb.npy' as NumPy binary array.")

```

```python
# Step 6: Save to .csv
df.to_csv('ambient_vib_pcb.csv', index=False)
print("[Saved] 'ambient_vib_pcb.csv' with sensor-channel labels.")
```
