```python
from nptdms import TdmsFile

# Load the TDMS file
tdms_file = TdmsFile.read("amb2.tdms")

# List all groups and channels
print("Groups and Channels:")
for group in tdms_file.groups():
    print(f"- Group: {group.name}")
    for channel in group.channels():
        print(f"  - Channel: {channel.name}")

# Access a specific channel (replace with actual group/channel names)
channel = tdms_file["amb2"]["502x"]
data = channel[:]

# Show basic info
print(f"\nChannel data length: {len(data)}")
print("First 10 values:", data[:10])
```
