```python
import scipy.io
import numpy as np
import pandas as pd

# Step 1: Load the .mat file
mat_data = scipy.io.loadmat('hammer_test.mat')
print("=== Variables in hammer_test.mat ===")
for key in mat_data:
    if not key.startswith("__"):
        print(f"{key}: {type(mat_data[key])}, shape: {np.shape(mat_data[key])}")

```

```python
# Step 2: Extract data arrays
hammerdata = mat_data['hammerdata']  # shape: (614400, 1)
tdata = mat_data['tdata']            # shape: (614400, 12)

```

```python
# Step 3: Define column names for tdata
sensor_labels = ['S1', 'S2', 'S3', 'S4']
axis_labels = ['X', 'Y', 'Z']
column_names = [f"{sensor}_{axis}" for sensor in sensor_labels for axis in axis_labels]

```

```python
# Step 4: Convert to DataFrames
df_tdata = pd.DataFrame(tdata, columns=column_names)
df_hammer = pd.DataFrame(hammerdata, columns=['Force'])

```

```python
# Step 5: Save hammerdata
np.save('free_vib_force.npy', hammerdata)
# df_hammer.to_csv('free_vib_force.csv', index=False)
print("[Saved] 'free_vib_force.npy'")

```

```python
# Step 6: Save tdata
np.save('free_vib_pcb.npy', tdata)
# df_tdata.to_csv('free_vib_pcb.csv', index=False)
print("[Saved] 'free_vib_pcb.npy'")
```
