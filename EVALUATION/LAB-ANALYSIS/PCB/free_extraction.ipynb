```python
import scipy.io
import numpy as np
import pandas as pd

# Step 1: Load the .mat file
mat_data = scipy.io.loadmat('hammertest2.mat')
print("=== Variables in hammertest2.mat ===")
for key in mat_data:
    if not key.startswith("__"):
        print(f"{key}: {type(mat_data[key])}, shape: {np.shape(mat_data[key])}")

```

```python
# Step 2: Extract data arrays
hammerdata = mat_data['hammerdata']  # shape: (368640, 1)
tdata = mat_data['tdata']            # shape: (368640, 27)

```

```python
# Step 3: Define column names for tdata
sensor_labels = ['S1', 'S2', 'S3', 'S4']
axis_labels = ['X', 'Y', 'Z']
column_names = [f"{sensor}_{axis}" for sensor in sensor_labels for axis in axis_labels]

selected_indices = [0, 1, 2, 3, 4, 5, 6, 7, 8, 12, 13, 14]
```

```python
# Step 4: Convert to DataFrames
df_tdata = pd.DataFrame(tdata[:, selected_indices], columns=column_names)
df_hammer = pd.DataFrame(hammerdata, columns=['Force'])

```

```python
# Step 5: Save DataFrame to .npy file
npy_tdata = df_tdata.to_numpy()
npy_hammer = df_hammer.to_numpy()

# check data_array shape
print("Data array shape:", npy_tdata.shape)
print("Hammer data shape:", npy_hammer.shape)
```

```python
# Step 6: Save tdata
np.save('free_vib_pcb.npy', npy_tdata)
np.save('free_vib_force.npy', npy_hammer)
```
